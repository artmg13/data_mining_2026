В качестве выбранного датасета выступает набор данных о пингвинах, находящийся по пути data/penguins.csv.

```python
# Импортируем необходимые библиотеки
import pandas as pd
```

```python
# Загружаем датасет и выводим первые строки
PATH = "data/penguins.csv"
df = pd.read_csv(PATH)
df.head(5)
```

```python
# Выведем информацию о датасете
df.info()
```

```python
# Вывод названия столбцов
print(df.columns)
```

```python
# Подсчитаем пропуски значений в строках
df.isna().sum()
```

**Выбрано удаление строк с пропущенными значениями, так как:**

1. Потеря данных небольшая: удаляются 11 из 344 строк (3,2%), остаются 333 наблюдения.
2. В столбце `sex` пропущено 11 значений. Это категориальный признак, поэтому заполнение медианой к нему неприменимо.
3. Заполнение числовых признаков общей медианой не учитывает различия между видами пингвинов и может создавать нехарактерные сочетания измерений. Удаление небольшого числа строк позволяет обойтись без внесения искусственных значений.


```python
# Удаляем пропуски и обновляем индексацию
df = df.dropna().reset_index(drop=True)

# Проверяем, что удаление было успешно
display(df.isna().sum())

# Выведем число оставшихся строк
print(f"Осталось строк: {len(df)}")
```

```python
# Проверим данные на дубликаты
df[df.duplicated()]
```
